## 4. Build corpus and search

Replace the URLs and keyword below, then run.

```python
urls = [
    "https://en.wikipedia.org/wiki/Digital_humanities",
    "https://en.wikipedia.org/wiki/Computational_linguistics",
]
keyword = "digital humanities"

corpus = ""
for url in urls:
    d = trafilatura.fetch_url(url)
    t = trafilatura.extract(d) or ""
    corpus += t
    hits = [s.strip() for s in t.split(".") if keyword.lower() in s.lower()]
    if hits:
        print("From", url[:40])
        [print("  ->", s) for s in hits[:2]]

open("corpus.txt","w").write(corpus)
print("Done:", len(corpus.split()), "words saved")
```
