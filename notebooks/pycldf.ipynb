## 3. Load a CLDF dataset

This cell downloads and loads a sample cross-linguistic dataset automatically.

```python
import pycldf, urllib.request, zipfile, pathlib
print('Downloading sample CLDF dataset...')
url = 'https://zenodo.org/records/13950591/files/cldf-datasets-wals-6c8a44c.zip'
urllib.request.urlretrieve(url, 'cldf.zip')
with zipfile.ZipFile('cldf.zip') as z: z.extractall('cldf_data')
ds = pycldf.Dataset.from_metadata(list(pathlib.Path('cldf_data').rglob('*.json'))[0])
print('Dataset loaded:', list(ds.components))
```

## 4. Analyse language distribution

```python
import pandas as pd, matplotlib.pyplot as plt, pathlib

# Load CSV files directly
csv_files = list(pathlib.Path('cldf_data').rglob('*.csv'))
print(f'Found {len(csv_files)} CSV files')

df = None
for f in csv_files:
    try:
        temp = pd.read_csv(f)
        if len(temp) > 10:
            df = temp
            print(f'Loaded: {f.name} — {len(df)} rows, {len(df.columns)} columns')
            print(df.head(3))
            break
    except:
        continue

if df is not None:
    col = df.columns[1]
    df[col].value_counts().head(10).plot(kind='bar', color='#534AB7')
    plt.title(f'Top 10 values in dataset')
    plt.tight_layout()
    plt.show()
else:
    print('No CSV data found')
```

## 5. Explore language groups

```python
if df is not None:
    print(f'Dataset summary: {len(df)} rows, {len(df.columns)} columns')
    print(f'Columns: {list(df.columns)}')
    col = df.columns[1]
    print(f'\nTop 5 values in {col}:')
    print(df[col].value_counts().head(5))
else:
    print('No data loaded.')
```
