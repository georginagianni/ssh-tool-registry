## 3. Load a CLDF dataset

This cell downloads and loads a sample cross-linguistic dataset automatically.

```python
import pycldf, urllib.request, zipfile, pathlib
print('Downloading sample CLDF dataset...')
url = 'https://zenodo.org/api/records/13149106/files/lexibank/allenbai-v4.1.zip/content'
urllib.request.urlretrieve(url, 'cldf.zip')
with zipfile.ZipFile('cldf.zip') as z: z.extractall('cldf_data')
ds = pycldf.Dataset.from_metadata(list(pathlib.Path('cldf_data').rglob('*.json'))[0])
print('Dataset loaded:', list(ds.components))
```

## 4. Analyse language distribution

```python
import pandas as pd, matplotlib.pyplot as plt

# Get available tables
print("Available tables:", list(ds.components))

# Try different table types
table = None
for t in ['LanguageTable', 'FormTable', 'EntryTable', 'ValueTable']:
    try:
        rows = list(ds[t])
        if rows:
            table = t
            df = pd.DataFrame(rows)
            print(f"Loaded {t}: {len(df)} rows")
            print(df.head(3))
            break
    except:
        continue

if df is not None:
    # Plot most common values in second column
    col = df.columns[1]
    df[col].value_counts().head(10).plot(kind='bar', color='#534AB7')
    plt.title(f'Top 10 values in {col}')
    plt.tight_layout()
    plt.show()
```

## 5. Explore language groups

```python
if df is not None:
    print(f"Dataset summary:")
    print(f"  Table: {table}")
    print(f"  Rows: {len(df)}")
    print(f"  Columns: {list(df.columns)}")
    # Show value distribution of most interesting column
    col = df.columns[1]
    print(f"\nTop 5 values in '{col}':")
    print(df[col].value_counts().head(5))
else:
    print("No table loaded.")
```
