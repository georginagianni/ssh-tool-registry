## 3. Load a CLDF dataset

This cell downloads and loads a sample cross-linguistic dataset automatically.

```python
import pycldf, urllib.request, zipfile, pathlib
print('Downloading sample CLDF dataset...')
url = 'https://zenodo.org/api/records/13149106/files/lexibank/allenbai-v4.1.zip'
urllib.request.urlretrieve(url, 'cldf.zip')
with zipfile.ZipFile('cldf.zip') as z: z.extractall('cldf_data')
ds = pycldf.Dataset.from_metadata(list(pathlib.Path('cldf_data').rglob('*.json'))[0])
print('Dataset loaded:', list(ds.components))
```

## 4. Analyse language distribution

```python
import pandas as pd, matplotlib.pyplot as plt
langs = list(ds['LanguageTable'])
df = pd.DataFrame(langs)
print(f'Total languages: {len(df)}')
print(df[['ID','Name','Glottocode']].head(10))
df['Name'].value_counts().head(10).plot(kind='bar', color='#534AB7')
plt.title('Top 10 languages in dataset')
plt.tight_layout()
plt.show()
```

## 5. Explore language groups

```python
if 'Family_ID' in df.columns:
    print(df['Family_ID'].nunique(), 'language families')
    print(df['Family_ID'].value_counts().head(5))
else:
    available = list(df.columns)
    print('Available columns:', available)
    print(df[available[1]].value_counts().head(5))
```
