## 5. Convert to all citation formats

```python
formats = ['bibtex','apalike','ris','codemeta']
for fmt in formats:
    r = subprocess.run(['cffconvert','--format',fmt], capture_output=True, text=True)
    with open(f'citation.{fmt}','w') as f: f.write(r.stdout)
    print(f'Done: {fmt}')
print('All formats saved.')
```
