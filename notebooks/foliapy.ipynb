## 3. Create a FoLiA document

Replace the sentences below with your own text, then run.

```python
import folia.main as folia

# Replace with your own sentences
sentences = [
    'Digital humanities research uses computational methods.',
    'Text annotation supports linguistic analysis.',
    'FoLiA is a standard format for annotated corpora.',
]

doc = folia.Document(id='my_document')
text = doc.add(folia.Text)
p = text.add(folia.Paragraph)
for sent in sentences:
    s = p.add(folia.Sentence)
    for word in sent.split(): s.add(folia.Word, word)
print('Words:', [w.text() for w in doc.words()])
print(f'Document created with {len(sentences)} sentences.')
```

## 4. Add POS annotations

Run this cell to automatically annotate your text with part-of-speech tags.

```python
doc.declare(folia.PosAnnotation, set='cgn-legacy')
pos_map = {
    'Digital': 'ADJ', 'humanities': 'N', 'research': 'N',
    'uses': 'V', 'computational': 'ADJ', 'methods': 'N',
    'Text': 'N', 'annotation': 'N', 'supports': 'V',
    'linguistic': 'ADJ', 'analysis': 'N',
    'FoLiA': 'N', 'is': 'V', 'a': 'ADJ', 'standard': 'N',
    'format': 'N', 'for': 'ADV', 'annotated': 'ADJ', 'corpora': 'N',
}
for word in doc.words():
    pos = pos_map.get(word.text(), 'N')
    word.add(folia.PosAnnotation, cls=pos)
doc.save('annotated.folia.xml')
count = len(list(doc.annotations(folia.PosAnnotation)))
print(f'Saved with {count} POS annotations to annotated.folia.xml')
```
