## 3. Create your RO-Crate

Fill in your name, institution and dataset title below, then run.

```python
from rocrate.rocrate import ROCrate
from rocrate.model.person import Person

# Fill in your details
your_name = "Your Name"
your_institution = "Your University"
dataset_title = "My Research Dataset"

crate = ROCrate()
crate.name = dataset_title
me = crate.add(Person(crate, 'https://orcid.org/0000-0000-0000-0000', {'name': your_name, 'affiliation': your_institution}))
crate.write('my_crate')
print(f'RO-Crate created for {your_name} at {your_institution}')
print('Metadata saved to my_crate/ro-crate-metadata.json')
```

## 4. Verify your crate

```python
import json
with open('my_crate/ro-crate-metadata.json', 'r') as f:
    crate_data = json.load(f)
print('=== RO-Crate Contents ===')
for entity in crate_data.get('@graph', []):
    etype = entity.get('@type', 'unknown')
    ename = entity.get('name', entity.get('@id', ''))
    print(f'  [{etype}] {ename}')
```

## 5. Add a research file

Upload any research file to include in your RO-Crate.

```python
import ipywidgets as widgets
from IPython.display import display
upload = widgets.FileUpload(multiple=False)
display(upload)
print('Upload a research file above.')
```

```python
if upload.value:
    val = list(upload.value.values())[0] if isinstance(upload.value, dict) else upload.value[0]
    fname = val['metadata']['name'] if isinstance(upload.value, dict) else val['name']
    raw = bytes(val['content'])
    with open(fname,'wb') as f: f.write(raw)
    crate.add_file(fname, properties={'name': fname, 'author': me})
    crate.write('my_crate')
    print(f'Added {fname} to RO-Crate.')
else:
    print('No file uploaded.')
```
